##### LOOP STATEMENTS

While Loop

A while loop in python iterates till its condition becomes False. In other words, it executes the statements under itself while the condition it takes is True. When the program control reaches the while loop, the condition is checked. If the condition is true, the block of code under it is executed. Remember to indent all statements under the loop equally. After that, the condition is checked again. This continues until the condition becomes false. Then, the first statement, if any, after the loop is executed.

```python
a = 0 #Assign the value 0 to the variable a

#The condition of this loop is to run *while* the value of a is LESS THAN 9
while(a < 9):
    print(a) #This print is to demonstrate how the condition functions in a loop using an number increment
    a += 1
   
print("Loop has ended at 8, can you tell why?")
```

##### For Loop

A for loop does the same thing as the example in the while loop, but makes it simpler to write and understand.

```python
# Get input from the user
num = int(input("Enter a positive integer: "))
```

```python
# Initialize the sum to 0
sum = 0

# Use a for loop to iterate from 1 to num (inclusive)
# The range() function generates a sequence of numbers
# starting from 1 and ending at num+1
for i in range(1, num+1):
    # Add the current number to the sum
    sum += i

# Display the result
print("The sum of numbers from 1 to", num, "is:", sum)
```

#### Array Indexing 

```python
a = np.array([[1,2], [3, 4], [5, 6]])

#Given the following, this is output of the above, can you see how you would get [1 4 5]? 
#For the value of '1' it is at position 0,0 which is the first 'box' and the first 'item' in the 'box'
#For the vlaue of '4' it is at the position of 1,1 which is the second 'box' and the second 'item' in the 'box'
#Can you now figure out how you would get the value of '5'?

print(np.array([a[0, 0], a[1, 1], a[2, 0]]))  # Prints "[1 4 5]"


#The above can also be written as:
# a[0,0] '1'
# a[1,1] '4'
# a[2,0] '5'
# This is an Example of integer array indexing. Look closely can you tell the difference between this statement and the last?
b=(a[[0, 1, 2], [0, 1, 0]]) # Do you now see how array indexing is done and how it relates to the previous statements?


print(b)  # Prints "[1 4 5]"
print(b.shape) #Shape is 3
```

```python
# The above example of integer array indexing is equivalent to this:
print(np.array([a[0, 0], a[1, 1], a[2, 0]]))  # Prints "[1 4 5]"

# When using integer array indexing, you can reuse the same
# element from the source array:
print(a[[0, 0], [1, 1]])  # Prints "[2 2]"

# Equivalent to the previous integer array indexing example
print(np.array([a[0, 1], a[0, 1]]))  # Prints "[2 2]"
```

```python
# Create an array of indices
b = np.array([1, 0, 1])
print(b)

# Select one element from each row of a using the indices in b
print(a[np.arange(3), b])
```

```python
# Mutate one element from each row of a using the indices in b
a[np.arange(3), b] += 10

print(a)
```

###### Array Mathematics​

```python
x = np.array([[1,2],[3,4]], dtype=np.float64)
y = np.array([[5,6],[7,8]], dtype=np.float64)

# Elementwise sum; both produce the array
# [[ 6.0  8.0]
#  [10.0 12.0]]
print(x + y)
print(np.add(x, y))
```

```python
X = np.array([[13.0,7.0],[4.0,47.0]], dtype=np.float64)
Y = np.array([[16.0,14.0],[1.0,23.0]], dtype=np.float64)
print (X-Y)
print(np.subtract(X, Y))
```

```python
#to perform elementwise multiplication
print(x * y)
print(np.multiply(x, y))
```

```python
#to perform elementwise division
print(x / y)
print(np.divide(x, y))
```
