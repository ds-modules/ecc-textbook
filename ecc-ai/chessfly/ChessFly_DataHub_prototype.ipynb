# 1. Training and inference

**Training** is the step where a network is shown examples and some of its numbers change. For ChessFly, the examples were chess positions with moves to learn. That work is finished. There is no training button on the board.

**Inference** is the step where those numbers stay fixed and the network is asked for an answer. When you play a move, ChessFly runs inference: it reads the board and returns a reply. Your game does not change the trained weights.
