# 1. From brain wiring to an AI model

A **connectome** is a map of nerve cells and the connections among them. FlyWire published that map for one adult female fruit fly: 138,639 neurons and 15,091,983 connections.

A **model architecture** is the layout of a computer model: which parts read the input, which parts pass signals along, and which parts produce the output. ChessFly's architecture follows the fly connectome. The connections stay where the map put them.

**Trained weights** are numbers learned from examples. In ChessFly, those numbers set how strongly each connection is used and how the chess board is translated into the network. The map says a connection exists. The weights say how strong it is.

**Inference** is using a finished model to answer a new question. The weights stay as they are. Playing a game is inference.

Here is the path of one chess position:

```text
chess position
    → input encoder
    → connectome-based network
    → output decoder
    → move scores
    → search chooses a move
```

The **input encoder** turns the board into signals on the fly's visual neurons. The **connectome-based network** passes those signals along the mapped connections for a few steps. The **output decoder** turns the resulting activity into a score for each legal move. **Search** then looks ahead at more than one possible future and picks the move the app will play.

**From FlyWire:** which neurons connect, and whether each connection is treated as excitatory or inhibitory.

**Added by the developers:** the encoder, the learned strengths, the decoder, the chess examples used in training, and the search that chooses the move you see.

ChessFly uses the female FlyWire brain. The assigned Google article is about a different map: one male fly's brain and nerve cord. Do not treat those as the same animal or the same dataset.

# 5. Training versus inference

Developers kept the wiring fixed and trained adjustable numbers using millions of chess examples. The examples were Stockfish-generated move and position evaluations for about 4.4 million Lichess positions. That training is already done. The model card is here: [mlabonne/chessfly](https://huggingface.co/mlabonne/chessfly).

**Optional note.** The adjustable numbers are an encoder from the board onto visual neurons, a positive strength on each connection, a gain and a threshold for each neuron at each step, and a decoder. A gain rescales a neuron's input. A threshold shifts when that neuron becomes active. The decoder reads central-brain neurons and descending neurons, which are cells with outputs toward the body, and turns their activity into move scores. You can play the game without memorizing this list.

The move on the board is not the decoder's score alone. The hosted app runs a search on top of those scores. A whole level of that search is one batched pass through the brain. When you see **positions evaluated**, you are seeing that search, not a second round of training.
